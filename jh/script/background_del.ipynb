```python
import json

json_path = "/home/elicer/dev/jh/data/train_40.json"

# JSON 불러오기
with open(json_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# -------------------------------------------------
# 1) background category(id=1) 제거
# -------------------------------------------------

background_id = 1

# annotations에서 background 제거
filtered_annotations = [
    anno for anno in data["annotations"]
    if anno["category_id"] != background_id
]

# -------------------------------------------------
# 2) categories에서 background 제거
# -------------------------------------------------

filtered_categories = [
    cat for cat in data["categories"]
    if cat["id"] != background_id
]

# -------------------------------------------------
# 3) category_id 재정렬 (1부터 다시)
# -------------------------------------------------

# 기존 id → 새 id 매핑 테이블 생성
old_to_new = {}
new_categories = []
new_id = 1

for cat in filtered_categories:
    old_to_new[cat["id"]] = new_id
    new_categories.append({
        "id": new_id,
        "name": cat["name"]
    })
    new_id += 1

# -------------------------------------------------
# 4) annotations의 category_id도 새 id로 매핑
# -------------------------------------------------

for anno in filtered_annotations:
    anno["category_id"] = old_to_new[anno["category_id"]]

# -------------------------------------------------
# 저장
# -------------------------------------------------

data["annotations"] = filtered_annotations
data["categories"] = new_categories

with open(json_path, "w") as f:
    json.dump(data, f, indent=4)

print("완료! background 제거 + 카테고리 재정렬 JSON 덮어쓰기 완료.")
```
